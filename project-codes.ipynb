```python

from datetime import datetime

# Class 1: Subscription
class Subscription:
    def __init__(self, name, amount, start_date):
        self.name = name
        self.amount = amount
        self.start_date = start_date
        self.status = "Active"

    def display(self):
        print("--------------------------------")
        print("Subscription Name:", self.name)
        print("Monthly Amount: Rs.", self.amount)
        print("Start Date:", self.start_date)
        print("Status:", self.status)

    def cancel(self):
        self.status = "Cancelled"
        print(self.name, "subscription cancelled successfully!")


# Class 2: Subscription Manager
class SubscriptionManager:
    def __init__(self):
        self.subscriptions = []

    # Add a new subscription
    def add_subscription(self):
        name = input("Enter subscription name: ")
        amount = float(input("Enter monthly amount: Rs. "))
        start_date = input("Enter start date (DD-MM-YYYY): ")

        sub = Subscription(name, amount, start_date)
        self.subscriptions.append(sub)

        print("Subscription added successfully!")

    # Display all subscriptions
    def show_subscriptions(self):
        if not self.subscriptions:
            print("No subscriptions found!")
        else:
            for sub in self.subscriptions:
                sub.display()

    # Calculate monthly expense
    def calculate_total(self):
        total = 0

        for sub in self.subscriptions:
            if sub.status == "Active":
                total += sub.amount

        print("Total Monthly Expense: Rs.", total)

    # Cancel a subscription
    def cancel_subscription(self):
        name = input("Enter subscription name to cancel: ")

        for sub in self.subscriptions:
            if sub.name.lower() == name.lower() and sub.status == "Active":
                sub.cancel()
                return

        print("Active subscription not found!")


# Main Program
manager = SubscriptionManager()

while True:
    print("\n===== SUBSCRIPTION MANAGEMENT SYSTEM =====")
    print("1. Add Subscription")
    print("2. View All Subscriptions")
    print("3. Calculate Monthly Expense")
    print("4. Cancel Subscription")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        manager.add_subscription()

    elif choice == "2":
        manager.show_subscriptions()

    elif choice == "3":
        manager.calculate_total()

    elif choice == "4":
        manager.cancel_subscription()

    elif choice == "5":
        print("Thank you for using the system!")
        break

    else:
        print("Invalid choice! Please try again.")
```
